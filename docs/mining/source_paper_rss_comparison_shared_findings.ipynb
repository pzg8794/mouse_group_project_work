# Source-paper RSS comparison shared findings

**Purpose.** This notebook summarizes the direct comparison between the source paper's RSS-prioritized AhR gene set and our local `mouse_new` DESeq2 + bend-point + GO/pathway-level workflow. It is written as a shareable findings notebook: minimal code, clear methods, figure-by-figure interpretation, and assets stored in this shared repository.

**Source paper / target gene set.** The source paper reports 1,431 AhR/RSS-prioritized genes from Supplementary Table 5, labelled as shifted genes in AhR cKO versus control DRGs after peripheral lesion ([Nature article](https://www.nature.com/articles/s41586-026-10295-z)). The paper's RSS quantity is `log2FC(cKO) - log2FC(control)`.

**Our local comparison.** We compare those source RSS genes against our DESeq2-tested universe, WT/FF and cKO/CRE side-specific contrasts, bend-point selected sets, direction of change, and retained GO membership outputs. The matching result should be read as **direct validation / comparison**, not as a claim that our pipeline exactly reproduces the source paper's original RSS workflow.

## Executive summary

- **What matched:** 1,025 of the 1,431 source-paper RSS genes were found in our DESeq2-tested gene universe.
- **Direction agreement:** 987 of those 1,025 matched source symbols agreed in RSS-like direction when comparing the source RSS sign with our local `cKO/CRE log2FC - WT/FF log2FC` delta.
- **Effect-size agreement:** source-vs-local log2FC values were strongly aligned: WT/control Pearson `r = 0.974`, cKO Pearson `r = 0.979`, and RSS-style delta Pearson `r = 0.941`.
- **Bend-point overlap:** a smaller subset entered the strict bend-point follow-up sets: 43 in WT/FF, 45 in cKO/CRE, 34 shared, 11 cKO-only, and 9 WT-only.
- **Source-constrained bend-point pass:** rerunning our bend-point rule inside the matched source RSS subset selected 37 WT/FF genes and 53 cKO/CRE genes, with 32 shared, 21 cKO-only, and 5 WT-only selected genes. The selected union contained 58 source RSS symbols, 55 of which were directionally consistent with the source RSS sign, and 46 selected gene IDs appeared in retained GO/pathway terms.
- **Biological interpretation:** the source-study stress/growth response-shift signal is largely present and directionally consistent in our tested universe, but our bend-point procedure intentionally narrows to the strongest local DESeq2 signal for pathway-level interpretation.

## Method overview: why this is apples-to-apples

**Question.** Do the source paper's AhR/RSS genes appear in our reanalysis, and do they behave consistently with our genotype-aware side-specific DESeq2 results?

**Method.**
1. Load source-paper Supplementary Table 5 and its RSS values.
2. Map source gene symbols to Ensembl gene IDs using source-paper Supplementary Tables 1 and 2, stripping Ensembl version suffixes.
3. Join mapped genes to our WT/FF and cKO/CRE DESeq2 side-specific result tables.
4. Compute a local RSS-like quantity: `our cKO/CRE log2FC - our WT/FF log2FC`.
5. Classify each source gene by DESeq2 significance, bend-point membership, branch specificity, direction agreement, and GO membership.
6. Apply the same bend-point rule inside the matched source RSS subset, separately for WT/FF and cKO/CRE, to test which source-paper genes survive our narrowing procedure.

**Why this method is fair.** The source paper used RSS to ask how the cKO injury response shifts relative to control. Our comparison asks the same gene-level question in our reanalysis, using the closest local analogue: the difference between the cKO/CRE side-specific injury contrast and the WT/FF side-specific injury contrast. The methods are not identical, so the comparison is a validation bridge rather than a reproduction claim.

**Why the source-constrained pass matters.** The first comparison is a membership and direction audit. The source-constrained pass is stricter: it starts with the source paper genes, reranks them using our DESeq2 p-values, applies the same bend-point rule, and then compares the selected source subset back to the full WT/FF and cKO/CRE branches.


## Why the analysis needed two steps

The two-step design is intentional and is the cleanest way to explain the comparison.

**Step 1: audit and validation bridge.** We first asked whether the source paper's RSS genes could be found and interpreted inside our local workflow. That step was necessary because the source paper and our project use related but different statistical objects: the source paper reports RSS genes, while our project uses DESeq2 side-specific contrasts, bend-point narrowing, and GO/pathway-level interpretation. The first pass therefore established the gene-identifier bridge, confirmed which source genes were present in our tested universe, checked whether source and local directions agreed, and showed where the source genes appeared in our existing bend-point and GO outputs.

**Step 2: source-constrained bend-point comparison.** After that bridge was established, we could ask the stricter apples-to-apples question: if the matched source RSS genes are passed through our own bend-point rule, which genes are selected in WT/FF and cKO/CRE? This second pass would not be defensible without Step 1 because we first needed to know that the genes were mapped correctly, present in our DESeq2 tables, and biologically/statistically comparable.

**Why this order matters.** Step 1 was exploratory in the scientific sense: it found the critical overlap, checked compatibility, and prevented a misleading comparison. Step 2 was confirmatory within our workflow: it applied the same bend-point selection logic directly to the validated source-gene subset.


## Methods and tools used

**Where the source genes came from.** The source-paper gene set comes from the Nature article's Supplementary Table 5, which reports 1,431 AhR/RSS-prioritized genes after peripheral lesion. Source gene-symbol to Ensembl identifier mapping was supported with the source paper's Supplementary Tables 1 and 2.

**Where our local statistics came from.** We used the local `mouse_new` DESeq2 outputs for the WT/FF and cKO/CRE ipsilateral-versus-contralateral contrasts. These tables provide the p-values, adjusted p-values, log2 fold changes, and base mean values used in the comparison. We also used the existing local bend-point selected sets, retained g:Profiler/GO membership tables, and anchor-gene tables to connect gene-level comparison back to pathway-level interpretation.

**Tools used.** The analysis was done in Python using `pandas` for table joins/counts, `numpy` for numerical calculations, `matplotlib` for the figures, and `openpyxl` to read the source-paper supplementary Excel workbook. The bend-point rule follows the same local implementation used in `mouse_new/mining/derive_de_analysis_all20.py`.

**Bend-point method in plain language.** For each branch, genes are ordered from smallest to largest p-value. The ordered p-value curve is transformed to `-log10(p-value)`, and the bend point is chosen as the gene farthest from the straight line connecting the start and end of the curve. Genes at or before that bend-point threshold are retained as the strongest local signal.

**Why this was the right approach.** The source paper's RSS measures how the cKO injury response shifts relative to control. Our closest reanalysis analogue is `cKO/CRE side-specific log2FC - WT/FF side-specific log2FC`. The first pass showed that this analogue is directionally compatible with the source RSS signal; the second pass then tested whether the source genes survive the same selection rule used throughout our project.

## Professor-facing explanation

A concise way to explain the workflow is:

> We did not jump straight to the source-constrained bend-point analysis because the source paper and our pipeline do not use the exact same statistical object. The source paper reports RSS genes, while our project uses DESeq2 side-specific contrasts followed by bend-point narrowing and GO/pathway interpretation. So the first pass was a necessary validation bridge: we mapped source symbols to Ensembl IDs, checked which genes were present in our tested universe, verified that source and local log2FC directions agreed, and identified where those genes appeared in our existing bend-point and GO outputs. Once that bridge was established, the second pass became the true method-level comparison: we took the matched source RSS genes and applied our same bend-point rule inside that source subset for WT/FF and cKO/CRE.
